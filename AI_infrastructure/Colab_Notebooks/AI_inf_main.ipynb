```python
import numpy as np

def filter_dataset(X, y, y_train):
    """
    Filters dataset (X, y) keeping only samples where for ALL attributes
    the label exists in y_train[attr].
    Assumes y and y_train are dicts with the same keys.
    """
    num_samples = len(X)
    keep_mask = np.ones(num_samples, dtype=bool)

    for attr in y_train:
        train_labels = set(y_train[attr])
        attr_labels = np.array(y[attr])
        attr_mask = np.array([lbl in train_labels for lbl in attr_labels])
        keep_mask = keep_mask & attr_mask

    # Filter X and y by the mask
    X_filtered = np.array(X)[keep_mask]
    y_filtered = {attr: np.array(y[attr])[keep_mask] for attr in y}

    return X_filtered, y_filtered


# Filter validation data
efnet_validation.X_val, efnet_validation.y_val = filter_dataset(
    efnet_validation.X_val,
    efnet_validation.y_val,
    efnet_training.y_train
)

# Filter testing data if exists
if efnet_testing.y_test is not None and efnet_testing.X_test is not None:
    if len(efnet_testing.X_test) > 0 and len(next(iter(efnet_testing.y_test.values()))) > 0:
        efnet_testing.X_test, efnet_testing.y_test = filter_dataset(
            efnet_testing.X_test,
            efnet_testing.y_test,
            efnet_training.y_train
        )
```
